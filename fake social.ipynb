```python
# STEP 33: HANDLE MISSING VALUES

# Create copies of the datasets
train_clean = train.copy()
test_clean = test.copy()

# Identify numerical and categorical columns
numeric_columns = train_clean.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_columns = train_clean.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Fill missing numerical values using the median
for column in numeric_columns:
    median_value = train_clean[column].median()
    train_clean[column] = train_clean[column].fillna(median_value)

    if column in test_clean.columns:
        test_clean[column] = test_clean[column].fillna(median_value)

# Fill missing categorical values using the most frequent value
for column in categorical_columns:
    if train_clean[column].notna().any():
        most_frequent = train_clean[column].mode()[0]
        train_clean[column] = train_clean[column].fillna(most_frequent)

        if column in test_clean.columns:
            test_clean[column] = test_clean[column].fillna(most_frequent)

print("Missing-value handling completed!")
```

```python
# STEP 63: SEPARATE FEATURES AND TARGET

# X contains the input features used to make predictions
X = train_clean.drop(columns=["fake"])

# y contains the correct account labels
y = train_clean["fake"]

# Display the shapes of X and y
print("Features shape:", X.shape)
print("Target shape:", y.shape)

# Display the number of fake and genuine accounts
print("\nAccount label distribution:")
print(y.value_counts().sort_index())

# Display the first five target labels
print("\nFirst five target labels:")
print(y.head())
```

```python
# STEP 64: SPLIT THE DATASET

from sklearn.model_selection import train_test_split

# Split the data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the sizes of the resulting datasets
print("Training features:", X_train.shape)
print("Validation features:", X_val.shape)

print("\nTraining labels:", y_train.shape)
print("Validation labels:", y_val.shape)

# Display label distribution in each set
print("\nTraining label distribution:")
print(y_train.value_counts().sort_index())

print("\nValidation label distribution:")
print(y_val.value_counts().sort_index())
```

```python
# STEP 65: TRAIN THE LOGISTIC REGRESSION MODEL

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Create the model
logistic_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)

# Train the model using the training data
logistic_model.fit(X_train, y_train)

# Display a success message
print("Logistic Regression model trained successfully!")
```

```python
# STEP 66: EVALUATE THE LOGISTIC REGRESSION MODEL

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Predict account labels using the validation data
y_pred_logistic = logistic_model.predict(X_val)

# Calculate model accuracy
accuracy_logistic = accuracy_score(y_val, y_pred_logistic)

print("LOGISTIC REGRESSION RESULTS")
print("===========================")

print("Accuracy:", round(accuracy_logistic * 100, 2), "%")

# Display precision, recall and F1-score
print("\nClassification Report:")
print(classification_report(y_val, y_pred_logistic, zero_division=0))

# Display the confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_val, y_pred_logistic))
```

```python
# STEP 67: TRAIN THE RANDOM FOREST MODEL

from sklearn.ensemble import RandomForestClassifier

# Create the Random Forest model
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

# Train the model
random_forest_model.fit(X_train, y_train)

# Display a success message
print("Random Forest model trained successfully!")
```

```python
# STEP 68: EVALUATE THE RANDOM FOREST MODEL

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Predict account labels using the validation dataset
y_pred_rf = random_forest_model.predict(X_val)

# Calculate accuracy
accuracy_rf = accuracy_score(y_val, y_pred_rf)

# Display the results
print("RANDOM FOREST RESULTS")
print("=====================")

print("Accuracy:", round(accuracy_rf * 100, 2), "%")

# Display precision, recall, and F1-score
print("\nClassification Report:")
print(classification_report(
    y_val,
    y_pred_rf,
    zero_division=0
))

# Display the confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_val, y_pred_rf))
```

```python
# STEP 69: COMPARE MODEL PERFORMANCE

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Calculate performance metrics for Logistic Regression
logistic_results = {
    "Model": "Logistic Regression",
    "Accuracy": accuracy_score(y_val, y_pred_logistic),
    "Precision": precision_score(y_val, y_pred_logistic, zero_division=0),
    "Recall": recall_score(y_val, y_pred_logistic, zero_division=0),
    "F1-Score": f1_score(y_val, y_pred_logistic, zero_division=0)
}

# Calculate performance metrics for Random Forest
rf_results = {
    "Model": "Random Forest",
    "Accuracy": accuracy_score(y_val, y_pred_rf),
    "Precision": precision_score(y_val, y_pred_rf, zero_division=0),
    "Recall": recall_score(y_val, y_pred_rf, zero_division=0),
    "F1-Score": f1_score(y_val, y_pred_rf, zero_division=0)
}

# Display the comparison table
comparison = pd.DataFrame([logistic_results, rf_results])

print("MODEL PERFORMANCE COMPARISON")
display(comparison.round(3))
```
